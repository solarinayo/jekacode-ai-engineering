```python
import os, requests, base64
import gradio as gr
from dotenv import load_dotenv

load_dotenv()
```

```python
MODELS = {
    "General Doctor AI": "openai/gpt-4o-mini",
    "Fast AI": "google/gemini-2.0-flash-001",
    "Reasoning AI": "deepseek/deepseek-chat",
}
```

```python
MODELS = {
    "Gemini": "google/gemini-2.0-flash-001",
    "GPT": "openai/gpt-4o-mini"
}
```

```python
SYSTEM = """
You are a cautious AI health-information assistant.
Provide general health information, possible next steps,
and safe self-care suggestions.

Do not claim to be a doctor.
Do not make definitive diagnoses or prescribe medication.
For emergencies, advise the user to seek urgent medical care.
"""
```

```python
def ask_ai(prompt, model):
    data = {
        "model": MODELS[model],
        "messages": [
            {"role": "system", "content": SYSTEM},
            {"role": "user", "content": prompt}
        ]
    }

    r = requests.post(
        URL,
        headers={"Authorization": f"Bearer {API_KEY}"},
        json=data,
        timeout=60
    )

    return r.json()["choices"][0]["message"]["content"]
```

```python
def consultation(symptoms, model):
    prompt = f"""
Symptoms/question:
{symptoms}

Give:
1. Possible general explanations
2. Safe self-care advice
3. Warning signs
4. When to see a healthcare professional
"""
    return ask_ai(prompt, model)
```

```python
def home_remedy(problem, model):
    prompt = f"""
The user asks about:
{problem}

Give only safe, general home-care measures.
Mention when home care is inappropriate and medical care is needed.
"""
    return ask_ai(prompt, model)
```

```python
def analyze_image(image, model):
    if image is None:
        return "Please upload an image."

    with open(image, "rb") as f:
        b64 = base64.b64encode(f.read()).decode()

    data = {
        "model": MODELS[model],
        "messages": [{
            "role": "user",
            "content": [
                {"type": "text", "text":
                 "Describe visible features in this image. "
                 "Do not diagnose. Give safe next steps and warning signs."},
                {"type": "image_url",
                 "image_url": {"url": f"data:image/jpeg;base64,{b64}"}}
            ]
        }]
    }

    r = requests.post(
        URL,
        headers={"Authorization": f"Bearer {API_KEY}"},
        json=data,
        timeout=90
    )

    return r.json()["choices"][0]["message"]["content"]
```

```python
def feedback(text, model):
    return ask_ai(
        f"Review this health-assistant response for clarity, safety and missing warnings:\n{text}",
        model
    )
```

```python
with gr.Blocks(title="AI Doctor Assistant") as app:

    gr.Markdown("# 🩺 Mr Olu 'THE' AI Doctor")
    gr.Markdown("General health information — not a replacement for a doctor.")

    model = gr.Dropdown(
        list(MODELS.keys()),
        value="General Doctor AI",
        label="AI Model"
    )

    with gr.Tab("Consultation"):
        symptoms = gr.Textbox(
            label="Describe your symptoms",
            lines=5
        )
        consult_btn = gr.Button("Consult")
        consult_out = gr.Markdown()

        consult_btn.click(
            consultation,
            [symptoms, model],
            consult_out
        )

    with gr.Tab("Home Care"):
        problem = gr.Textbox(
            label="What do you need help with?",
            lines=4
        )
        remedy_btn = gr.Button("Get Advice")
        remedy_out = gr.Markdown()

        remedy_btn.click(
            home_remedy,
            [problem, model],
            remedy_out
        )

    with gr.Tab("Image Analysis"):
        image = gr.Image(
            type="filepath",
            label="Upload health-related image"
        )
        image_btn = gr.Button("Analyze")
        image_out = gr.Markdown()

        image_btn.click(
            analyze_image,
            [image, model],
            image_out
        )

    with gr.Tab("Feedback"):
        response = gr.Textbox(
            label="Paste AI response",
            lines=7
        )
        feedback_btn = gr.Button("Review")
        feedback_out = gr.Markdown()

        feedback_btn.click(
            feedback,
            [response, model],
            feedback_out
        )
```
