```python
import os
import requests
import gradio as gr
from dotenv import load_dotenv

load_dotenv()

OPENROUTER_API_KEY = os.environ.get("OPENROUTER_API_KEY")

if not OPENROUTER_API_KEY:
    raise ValueError("OPENROUTER_API_KEY was not found. Check your .env file.")
```

```python
OPENROUTER_URL = "https://openrouter.ai/api/v1/chat/completions"

HEADERS = {
    "Authorization": f"Bearer {OPENROUTER_API_KEY}",
    "Content-Type": "application/json"
}
```

```python
MODEL = "openrouter/free"
```

```python
SYSTEM_PROMPT = """
You are FIRSTGATE's professional customer service representative.

Your responsibilities are:

1. Answer customers politely and professionally.
2. Understand the customer's question before responding.
3. Give clear, useful and concise answers.
4. If you do not know something, say so rather than inventing information.
5. Help customers with general questions, complaints, orders, payments,
   refunds and other customer-service matters.
6. Remain calm and respectful when customers are frustrated.
7. Never claim that an action has been completed if you cannot actually
   perform that action.
8. Escalate serious or complicated issues to a human representative.

Company name:
FIRSTGATE

Always respond as a helpful FIRSTGATE customer service representative.
"""
```

```python
def customer_service(message, history):
    
    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        }
    ]

    # Add previous conversation
    for user_message, assistant_message in history:
        messages.append({
            "role": "user",
            "content": user_message
        })

        messages.append({
            "role": "assistant",
            "content": assistant_message
        })

    # Add current customer message
    messages.append({
        "role": "user",
        "content": message
    })

    payload = {
        "model": MODEL,
        "messages": messages
    }

    try:
        response = requests.post(
            OPENROUTER_URL,
            headers=HEADERS,
            json=payload,
            timeout=60
        )

        response.raise_for_status()

        data = response.json()

        return data["choices"][0]["message"]["content"]

    except requests.exceptions.RequestException as e:
        return f"Sorry, there was a connection problem: {e}"

    except (KeyError, IndexError):
        return "Sorry, I could not process the AI response."
```

```python
demo = gr.ChatInterface(
    fn=customer_service,
    title="FIRSTGATE Customer Service Portal",
    description="AI-powered customer service representative",
    textbox=gr.Textbox(
        placeholder="Type your customer message here...",
        label="Customer Message"
    )
)
```

```python
demo.launch()
```
