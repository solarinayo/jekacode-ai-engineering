# AI Writing Assistant — 3 free models via OpenRouter

One API, three free models compared side by side. Modes: **summarize**, **create**, **rewrite**.

**Setup before running:**
- `pip install requests python-dotenv`
- `.env` file (same folder as this notebook) with:
  ```
  OPENROUTER_API_KEY=your_openrouter_key
  ```
- Get a free key at https://openrouter.ai/keys — no card needed.


````
import os
import requests
from dotenv import load_dotenv

load_dotenv()
OPENROUTER_API_KEY = os.environ.get("OPENROUTER_API_KEY")

print("OpenRouter key loaded:", OPENROUTER_API_KEY is not None)
````

## The three free models to compare

All routed through OpenRouter's single OpenAI-compatible endpoint. Free model IDs rotate over time — check https://openrouter.ai/collections/free-models if one of these ever stops responding.

````
MODELS = {
    "GPT-OSS-120B": "openai/gpt-oss-120b:free",
    "QWEN3-CODER": "qwen/qwen3-coder:free",
    "GEMMA-4": "google/gemma-4-31b-it:free",
}
````

## The single model caller

````
def ask_openrouter(prompt, model):
    url = "https://openrouter.ai/api/v1/chat/completions"
    headers = {
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "HTTP-Referer": "https://jekacode.africa",  # OpenRouter asks for this, any URL works
        "X-Title": "AI Writing Assistant",
    }
    payload = {"model": model, "messages": [{"role": "user", "content": prompt}]}
    r = requests.post(url, headers=headers, json=payload, timeout=30)
    r.raise_for_status()
    return r.json()["choices"][0]["message"]["content"].strip()
````

## Prompt builder for the 3 writing modes

````
def build_prompt(mode, text):
    if mode == "summarize":
        return f"Summarize the following text in 2-3 clear sentences:\n\n{text}"
    if mode == "create":
        return f"Write a short, engaging piece of content based on this idea:\n\n{text}"
    if mode == "rewrite":
        return f"Rewrite the following text to be clearer and more engaging, keeping the same meaning:\n\n{text}"
    raise ValueError("mode must be summarize, create, or rewrite")
````

## Run one prompt through all 3 models

````
def run_all_models(prompt):
    for name, model_id in MODELS.items():
        print(f"\n--- {name} ---")
        try:
            print(ask_openrouter(prompt, model_id))
        except Exception as e:
            print(f"[{name} failed: {e}]")
````

## Try it — edit `mode` and `text` and re-run this cell

````
mode = "summarize"  # summarize / create / rewrite
text = "Paste the text you want summarized, created from, or rewritten here."

prompt = build_prompt(mode, text)
run_all_models(prompt)
````
