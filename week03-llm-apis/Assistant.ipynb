```python
from huggingface_hub import InferenceClient
```

```python
groq_client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

genai.configure(
    api_key=os.getenv("GEMINI_API_KEY")
)

hf_client = InferenceClient(
    token=os.getenv("HF_TOKEN")
)
```

```python
def groq_model(task, content):
    prompt = f"""
    Task: {task}

    Content:
    {content}
    """

    response = groq_client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
```

```python
def gemini_model(task, content):
    prompt = f"""
    Task: {task}

    Content:
    {content}
    """

    model = genai.GenerativeModel(
        os.getenv("GEMINI_MODEL")
    )

    response = model.generate_content(prompt)

    return response.text
```

```python
def huggingface_model(task, content):
    prompt = f"""
    Task: {task}

    Content:
    {content}
    """

    response = hf_client.text_generation(
        prompt,
        model="HuggingFaceH4/zephyr-7b-beta",
        max_new_tokens=300
    )

    return response
```

```python
content = """
Artificial intelligence is changing the way businesses analyze data,
communicate with customers, and automate repetitive tasks. Organizations
are increasingly using AI to improve efficiency and make better decisions.
"""

generate_task = """
Generate a clear and informative paragraph based on the topic.
"""

summarize_task = """
Summarize the content into a short paragraph while keeping the main ideas.
"""

rewrite_task = """
Rewrite the content in a clearer and more professional way without
changing its meaning.
"""
```

```python
print("===== GROQ =====")
print(groq_model(generate_task, content))

print("\n===== GEMINI =====")
print(gemini_model(generate_task, content))

print("\n===== HUGGING FACE =====")
print(huggingface_model(generate_task, content))
```
