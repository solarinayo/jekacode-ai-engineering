```python
import os
import requests
import gradio as gr

from dotenv import load_dotenv


load_dotenv()

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

OPENROUTER_URL = "https://openrouter.ai/api/v1/chat/completions"

print("OpenRouter key loaded:", bool(OPENROUTER_API_KEY))
```

```python
def ask_openrouter(messages, model="openrouter/free"):
    """
    Send a request to OpenRouter and return the response.
    """

    if not OPENROUTER_API_KEY:
        return "OpenRouter API key was not found."

    headers = {
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "HTTP-Referer": "https://jekacode.africa",
        "X-Title": "AI Writing Assistant",
        "Content-Type": "application/json"
    }

    payload = {
        "model": model,
        "messages": messages,
        "max_tokens": 700
    }

    try:
        response = requests.post(
            OPENROUTER_URL,
            headers=headers,
            json=payload,
            timeout=60
        )

        if response.status_code == 429:
            return "The model is temporarily rate-limited. Please try again."

        if response.status_code == 403:
            return "The selected model is currently unavailable. Please try again."

        response.raise_for_status()

        result = response.json()

        return result["choices"][0]["message"]["content"].strip()

    except requests.Timeout:
        return "The request took too long. Please try again."

    except requests.RequestException as error:
        return f"API request error: {error}"

    except (KeyError, IndexError):
        return "The API returned an unexpected response format."

    except Exception as error:
        return f"Unexpected error: {error}"
```

```python
def writing_assistant(task, text):
    """
    Process the user's writing request.
    """

    if not text or not text.strip():
        return "Please enter some text or instructions."

    prompt = f"""
You are an AI Writing Assistant.

Writing task:
{task}

User's text or instruction:
{text}

Complete the task clearly and professionally.
Preserve the original meaning where appropriate.
Return only the improved or generated writing unless a short explanation is necessary.
"""

    return ask_openrouter(
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        model="openrouter/free"
    )
```

```python
with gr.Blocks(title="AI Writing Assistant") as writing_demo:

    gr.Markdown(
        """
        # ✍️ AI Writing Assistant

        Write, rewrite, summarize, and improve your content.
        """
    )

    task = gr.Dropdown(
        choices=[
            "Write new content",
            "Rewrite text",
            "Improve grammar",
            "Summarize text",
            "Make text more professional",
            "Make text shorter",
            "Create an email",
            "Create a social media caption"
        ],
        value="Rewrite text",
        label="Choose a writing task"
    )

    text_input = gr.Textbox(
        label="Enter your text or instructions",
        placeholder="Type or paste your text here...",
        lines=8
    )

    generate_button = gr.Button(
        "Generate Writing",
        variant="primary"
    )

    output = gr.Markdown(
        label="Assistant Response"
    )

    generate_button.click(
        fn=writing_assistant,
        inputs=[task, text_input],
        outputs=output
    )


if __name__ == "__main__":
    writing_demo.launch()
```
