```python
students = [
    {"name": "David", "score": 72},
    {"name": "Sarah", "score": 65},
    {"name": "John", "score": 48},
    {"name": "Mary", "score": 81},
    {"name": "Daniel", "score": 55},
    {"name": "Grace", "score": 39}
]

def get_class_stats(students):
    scores = [student["score"] for student in students]

    average = sum(scores) / len(scores)
    highest = max(scores)
    lowest = min(scores)

    return average, highest, lowest

# Calculate class statistics
average, highest, lowest = get_class_stats(students)

print(f"Class average: {average:.2f}")
print(f"Highest score: {highest}")
print(f"Lowest score: {lowest}")

print("\nStudent Results:")

#Print pass/fail for each student
for student in students:
    if student["score"] >= 50:
        print(f"{student['name']}: Pass")
    else:
        print(f"{student['name']}: Fail")


#Stretch
if "ask_gemini_safely" in globals():
    prompt = f"""
Write a short paragraph summarizing this class.

Class average: {average:.2f}
Highest score: {highest}
Lowest score: {lowest}
"""

    summary = ask_gemini_safely(prompt)
    print("\nClass Summary")
    print(summary)
else:
    print("\nGemini summary skipped: ask_gemini_safely() is not available.")

```
