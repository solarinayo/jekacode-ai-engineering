```python
import os
import requests
from dotenv import load_dotenv

# Load the .env file from the project root
load_dotenv()

print("Environment check:")
print("Gemini key:", bool(os.getenv("GEMINI_API_KEY")))
print("Groq key:", bool(os.getenv("GROQ_API_KEY")))
print("OpenRouter key:", bool(os.getenv("OPENROUTER_API_KEY")))

print("\nModels:")
print("Gemini model:", os.getenv("GEMINI_MODEL"))
print("Groq model:", os.getenv("GROQ_MODEL"))
```

```python
def ask(provider, prompt):

    # =========================
    # GEMINI
    # =========================
    if provider == "gemini":

        model = os.getenv("GEMINI_MODEL")

        url = f"https://generativelanguage.googleapis.com/v1beta/models/{model}:generateContent"

        headers = {
            "x-goog-api-key": os.getenv("GEMINI_API_KEY"),
            "Content-Type": "application/json"
        }

        body = {
            "contents": [
                {
                    "parts": [
                        {
                            "text": prompt
                        }
                    ]
                }
            ]
        }

    # =========================
    # GROQ
    # =========================
    elif provider == "groq":

        model = os.getenv("GROQ_MODEL")

        url = "https://api.groq.com/openai/v1/chat/completions"

        headers = {
            "Authorization": f"Bearer {os.getenv('GROQ_API_KEY')}",
            "Content-Type": "application/json"
        }

        body = {
            "model": model,
            "messages": [
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        }

    # =========================
    # OPENROUTER
    # =========================
    elif provider == "openrouter":

        url = "https://openrouter.ai/api/v1/chat/completions"

        headers = {
            "Authorization": f"Bearer {os.getenv('OPENROUTER_API_KEY')}",
            "Content-Type": "application/json"
        }

        body = {
            "model": "openrouter/free",
            "messages": [
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        }

    else:
        raise ValueError(f"Unknown provider: {provider}")

    # Send request
    response = requests.post(
        url,
        headers=headers,
        json=body,
        timeout=60
    )

    # Show useful error information
    if not response.ok:
        print("\n" + "=" * 60)
        print(f"{provider.upper()} API ERROR")
        print("=" * 60)
        print("Status code:", response.status_code)
        print("Response:")
        print(response.text[:2000])
        response.raise_for_status()

    # Convert response to JSON
    data = response.json()

    # Extract response text
    if provider == "gemini":
        return data["candidates"][0]["content"]["parts"][0]["text"]
    else:
        return data["choices"][0]["message"]["content"]
```

```python
print("Testing Gemini...")
print("=" * 60)

result = ask(
    "gemini",
    "Write one short sentence about learning artificial intelligence."
)

print(result)
```

```python
def ask(provider, prompt):

    # =========================
    # GEMINI
    # =========================
    if provider == "gemini":

        model = os.getenv("GEMINI_MODEL")

        url = f"https://generativelanguage.googleapis.com/v1beta/models/{model}:generateContent"

        headers = {
            "x-goog-api-key": os.getenv("GEMINI_API_KEY"),
            "Content-Type": "application/json"
        }

        body = {
            "contents": [
                {
                    "parts": [
                        {
                            "text": prompt
                        }
                    ]
                }
            ]
        }


    # =========================
    # GROQ
    # =========================
    elif provider == "groq":

        model = os.getenv("GROQ_MODEL")

        url = "https://api.groq.com/openai/v1/chat/completions"

        headers = {
            "Authorization": f"Bearer {os.getenv('GROQ_API_KEY')}",
            "Content-Type": "application/json"
        }

        body = {
            "model": model,
            "messages": [
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        }


    # =========================
    # OPENROUTER
    # =========================
    elif provider == "openrouter":

        url = "https://openrouter.ai/api/v1/chat/completions"

        headers = {
            "Authorization": f"Bearer {os.getenv('OPENROUTER_API_KEY')}",
            "Content-Type": "application/json"
        }

        body = {
            "model": "openrouter/free",
            "messages": [
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        }


    else:
        raise ValueError(f"Unknown provider: {provider}")


    # =========================
    # SEND REQUEST
    # =========================

    response = requests.post(
        url,
        headers=headers,
        json=body,
        timeout=60
    )


    # =========================
    # ERROR HANDLING
    # =========================

    if not response.ok:

        print("\n" + "=" * 60)
        print(f"{provider.upper()} API ERROR")
        print("=" * 60)
        print("Status code:", response.status_code)
        print("Response:")
        print(response.text[:2000])

        response.raise_for_status()


    data = response.json()


    # =========================
    # RETURN RESPONSE
    # =========================

    if provider == "gemini":

        return data["candidates"][0]["content"]["parts"][0]["text"]

    else:

        return data["choices"][0]["message"]["content"]
```

```python
models = ["gemini", "groq", "openrouter"]

print("Models being tested:")
for model in models:
    print("-", model)
```

```python
print("GENERATE")
print("=" * 60)

for model in models:
    print(f"\n{model.upper()}")
    print("-" * 60)

    prompt = """
Write a short paragraph about why learning artificial intelligence
is important for students.
"""

    result = ask(model, prompt)

    print(result)
```

```python
print("SUMMARIZE")
print("=" * 60)

for model in models:
    print(f"\n{model.upper()}")
    print("-" * 60)

    prompt = f"""
Summarize the following content in one clear sentence:

{content}
"""

    result = ask(model, prompt)

    print(result)
```

```python
print("REWRITE")
print("=" * 60)

for model in models:
    print(f"\n{model.upper()}")
    print("-" * 60)

    prompt = f"""
Rewrite the following content in a professional tone:

{content}
"""

    result = ask(model, prompt)

    print(result)
```

```python
print("COMPARISON OF AI PROVIDERS")
print("=" * 60)

for model in models:
    print(f"\n{model.upper()}")
    print("-" * 60)

    generate_prompt = """
Write a short paragraph about why learning artificial intelligence
is important for students.
"""

    summarize_prompt = f"""
Summarize the following content in one clear sentence:

{content}
"""

    rewrite_prompt = f"""
Rewrite the following content in a professional tone:

{content}
"""

    generated = ask(model, generate_prompt)
    summarized = ask(model, summarize_prompt)
    rewritten = ask(model, rewrite_prompt)

    print("\nGENERATED:")
    print(generated)

    print("\nSUMMARY:")
    print(summarized)

    print("\nREWRITE:")
    print(rewritten)
```
